## python lambda function
### lambda fucntion is small anonymous fucntion (a function without formal name)

```python
#syntax lambda function
lambda arguments : expression 
```

```python
#noraml function
def add (a, b):
    return (a+b)
result = add(10, 20)
print (result)
```

```python
add = lambda a, b: a+b #here a, b are parameter , a+b is expression
result=add(10, 20) # arguments 10, 20
print (result)

#add variable name to stire the lambda fucntion
#lambda is keyword creating anonymous function
#a,b are parameter 
#a+b is expression

#Note: for lambda not required any return keyword becuase lambda automatically resturen expresisoin result 

```

```python
veera = lambda x: x * x
print(veera(5))
```

```python
check = lambda x:  "even" if x % 2 == 0 else "odd"
print(check(10))
print(check(7))
```

```python
check = lambda x: "positive" if x > 0 == 0 else "negative"
print(check(-7))
print(check(10))

#"positive" if x > 0 == 0 else "negative" --> expression
```

## Walrus Opeartors 

#### walrus opeartor := is used to value to variable use same variabel i same expresison

```python
#without walrus operator 
numbers = [10,20,30,40]
count = len(numbers)
if count > 3 :
    print (count)
```

```python
#with walrus operator
numbers = [10,20,30,40]

if (count:= len(numbers))> 3:
    print (count)
```

```python
if (num := eval(input("enter number")))>10:
    print ("number is greater than 10")
else:
    print("numebr is less than 10")
```

```python
check = lambda x:x > 10
if check(num := eval(input("enter number"))):
    print ("number is greater than 10")
else:
    print("numebr is less than 10")
```
