## 1.3 Python function : With parmeter and return value 
### Accept input argument from the function call
### Return keyword give result 
### This retrun keyword we can pass input to another function 

```python
def veera(a, b): # here a, b parameter 
    return a+b   # return keyword give result 
result=veera(10, 20)     # 10, 20 input argument to function call 
print(result)
#note we can use this return value to another function if reuired 
```

```python
def caliculate_bill (price, quantity, discount):  #price 1000 # quantity 3 #discount  10
    total = price * quantity
    discount_amount = total * discount / 100
    final_amount = total - discount_amount
    return final_amount   #i need final_amount as result
caliculate_bill(2, 2000, 20)
```

```python
## example multiple values as return 
def caliculate_bill (price, quantity, discount):  #price 1000 # quantity 3 #discount  10
    total = price * quantity
    discount_amount = total * discount / 100
    final_amount = total - discount_amount
    return final_amount,discount_amount, total   #to call multiple values 
caliculate_bill(1000, 3, 10)
```

## Function with Local and global variable 
### global variable : define outside of the function 
### local variable : define inside of the function

```python
def naresh ():
    name = "veera" # inside of the function local variable 
    print ("hello", name)
naresh()
```

```python
name = "FDE"
def naresh ():
    name = "veera" # inside of the function local variable 
    print ("hello", name) # print local variable
naresh()
print(name) #print global variable 
```

```python
salary = 50000
def update_salary():
    global salary #this global block tells python Dont create new local varible use existing variables name salary 
    salary= 60000
final=update_salary()
print(salary)

#intially salary is = 50000
#global block tells python to use the global variable so its change to 60000
```
