## 1.3 Default arguments 
#### Default arguments are used when the caller does not provide a value for that parameter.

```python
def student(name, age=25, city="Hyderabad"):
    print(name)
    print(age)
    print(city)

student("Veera")
student("Rahul", city="Bangalore")
student("Akash", 22, "Chennai")
```

## 1.4.1 Variable-Length Positional Arguments (*args)
#### allow a function to accept any number of positional arguments without defining each parameter separately.

```python
def student(*names):
    print(names)

student("Veera", "Rishabh", "Akash", "Rahul", "Kiran", "FDE")
```

## 1.4.2 Keyword-Only Arguments
### Keyword-only arguments are the arguments that must be passed using their parameter names when calling a function. We cannot pass them using positional values

```python
def function(a, *, b, c):
    pass
| Parameter | How can we pass values? | Example |
|---|---|---|
| `a` | Position or keyword | `10` or `a=10` |
| `b` | Only keyword | `b=20` |
| `c` | Only keyword | `c=30` |
```

```python
def employee(name, *, salary, city):
    print(name)
    print(salary)
    print(city)

employee("Veera", salary=50000, city="Hyderabad")
```

```python
def employee(name, salary, city):
    print(name, salary, city)

employee("Veera", 50000, "Hyderabad")  # Valid
employee("Veera", salary=50000, city="Hyderabad")  # Valid
```

```python
def employee(name, *, salary, city):
    print(name, salary, city)

employee("Veera", salary=50000, city="Hyderabad")  # Valid
```

```python
employee("Veera", 50000, "Hyderabad")  # Error
## becuse must use key-word only argument only 
```

## 1.4.3 Variable-Length Keyword Arguments (**kwargs)
### allow a function to accept any number of keyword arguments without defining each parameter separately.
#### **kwargs collects multiple keyword arguments into a dictionary.
#### The name kwargs is just a convention. We can use any valid name after **

```python
def student(**details):
    print(details)

student(name="Veera", age=30, city="Hyderabad")
```

```python
def student (name,  *marks, **details):
    print(name)
    print(marks)
    print(details)
student("veera", 
        80,
        90,
        100,
        city = "hyd",
         course= "fde"
       )
*marks = collect many val
```

```python
def student(*name, details, **marks):
    print(name)
    print(marks)
    print(details)

student(
    "veera",
    "rishabh",
    details="hyd",
    physics=80,
    chemistry=90,
    age=30
)
```

**Write a Python function to manage student registration using different types of function parameters**.

Requirements:
1. Every student must provide one roll number.
2. A student can select multiple subjects (e.g., Python, AWS, Azure).
3. The student must provide their city.
4. The student can provide additional details such as name, age, course, height, etc.
5. Display all the student information.

```python
## 1.Standard order
def student(rollno, *subjects, city, **details):
    print(rollno, subjects, city, details)

student(101, "Python", "AWS",
        city="Hyderabad", name="Veera")
```

**1. `rollno` – Normal Parameter**
- Accepts one value (positional or keyword).

**2. `*subjects` – Variable-Length Positional Arguments**
- Accepts multiple positional values and collects them into a tuple.

**3. `city` – Keyword-Only Parameter**
- Accepts one value using its parameter name.

**4. `**details` – Variable-Length Keyword Arguments**
- Accepts multiple keyword values and collects them into a dictionary.

```python
## 2. Swap rollno and *subjects (valid)
def student(*subjects, rollno, city, **details):
    print(rollno, subjects, city, details)

student("Python", "AWS",
        rollno=101, city="Hyderabad", name="Veera")
```

```python
## 3. Move city before *subjects (valid)
def student(rollno, city, *subjects, **details):
    print(rollno, subjects, city, details)

student(101, "Hyderabad", "Python", "AWS",
        name="Veera")
#Why valid? city becomes a normal positional-or-keyword parameter.
```

```python
## 4. Swap rollno and city after *subjects (valid)
def student(city,  rollno, *subjects, **details):
    print(rollno, subjects, city, details)

student("Python", "AWS",
        "Hyderabad", 101, name="Veera")
```

```python
## 6. Put rollno directly after **details (invalid)
# SyntaxError
def student(*subjects, city, **details, rollno):
    print(rollno, subjects, city, details)
studnet (
    "aws",
    "azure",
    "hyderabad",
    name = "veera",
    height = 172,
    32)
```

```python
SWAPPING RULES – PYTHON FUNCTION PARAMETERS

1. rollno, *subjects, city, **details  → VALID
2. *subjects, rollno, city, **details  → VALID
3. rollno, city, *subjects, **details  → VALID
4. *subjects, city, rollno, **details  → VALID
5. rollno, *subjects, **details, city  → INVALID
6. *subjects, city, **details, rollno  → INVALID

IMPORTANT RULES:

1. Normal parameters can come before *args.  
2. Parameters after *args become keyword-only.
3. Keyword-only parameters can be swapped.
4. **kwargs must always be the last parameter.
5. No parameter can be placed after **kwargs.
```

## Function with Local and global variable 
### global variable : define outside of the function 
### local variable : define inside of the function

```python
def naresh ():
    name = "veera" # inside of the function local variable 
    print ("hello", name)
naresh()
```

```python
name = "FDE"
def naresh ():
    name = "veera" # inside of the function local variable 
    print ("hello", name) # print local variable
naresh()
print(name) #print global variable 
```

```python
salary = 50000
def update_salary():
    global salary #this global block tells python Dont create new local varible use existing variables name salary 
    salary= 60000
final=update_salary()
print(salary)

#intially salary is = 50000
#global block tells python to use the global variable so its change to 60000
```
