## Instance Variables
### An instance variable is a variable that belongs to an individual object (instance) of a class.

```python
# example class variables and instance variable 
class Student:
    school = "ABC" #class variable as deffines inside the class
    country = "India"
    #insatance method
    def display(self, name, roll, age): #self is a reference to the current instance of the class and is used to access variables that belongs to the class.
      self.name = name #instance variable as deffines inside the fucntion method
      self.roll = roll
      self.age = age
s1=Student() # creation of object of class Student
s1.display("veera", 232, 20)
#class variable can be accessed using object of the class
print(s1.school) #accessing class variable using object of class Student
print(s1.country) #accessing class variable using object of class Student

##instance varibles can be accessed using object of the class
print(s1.name) #accessing instance variable using object of class Student
print(s1.roll) #accessing instance variable using object of class Student
print(s1.age) #accessing instance variable using object of class Student

```

```python
#with  __init__ constructor #it is special method
class Student:
    country = "Aus" #class variable as deffines inside the class
    def __init__(self, name, roll, age):
        self.name = name
        self.roll = roll
        self.age = age

s1 = Student("veera", 232, 20)
print(s1.name) #accessing instance variable using object of class Student
print(s1.roll)
print(s1.age)
print(s1.country) #accessing class variable using object of class Student

# __init__ is a special method.
# It is automatically executed when an object is created.
# It initializes the object's attributes.
# self refers to the current object.
# name, roll, and age are parameters.
```

## Encapsulation 
## Encapsulation helps protect an object's data from accidental direct modification by providing controlled ways to access or modify it.

```python
#without encapsulation#
class Bank:
    bank_name = "HDFC"
    def __init__(self, name, balance):
        self.name = name
        self.balance= balance
    def show_balance(self):
        print("Name: ", self.name)
        print("Balance: ", self.balance)
    def deposit(self, amount):
        self.balance+= amount
        print("Deposit: ", amount)
    def withdraw(self, amount):
            self.balance-= amount
            print("withdraw: ", amount)
        
b1 = Bank("veera", 10000)
b1.deposit(5000)
b1.withdraw(5000)
b1.balance = 0 ## Directly modifying balance from outside
b1.show_balance()

    
```

```python
#with encapsulation#
#protect data inside the class not form outside the class
# we can control our data through inside the class 
#helps data safce and control secure 
class Bank:
    bank_name = "HDFC"
    def __init__(self, name, balance):
        self.name = name
        self.__balance= balance
    def show_balance(self):
        print("Name: ", self.name)
        print("Balance: ", self.__balance)
    def deposit(self, amount):
        self.__balance+= amount
        print("Deposit: ", amount)
    def withdraw(self, amount):
            self.__balance-= amount
            print("withdraw: ", amount)
            
        
b1 = Bank("veera", 10000)
b1.show_balance()
b1.deposit(5000)
b1.withdraw(5000)
b1.__balance = 0
b1.show_balance()

## here __balance is private variable 
```

## 2.Inheritance method is an oops conecpt that allows one class to aquire the properties and methods of another class 

```python
# without inheritance #
# separate classes Parent and Child #
class Parent: #create class 
    def house(self):
        print("parent own house")
    def car(self):
        print("parent own car")
class Child:
    def bike(self):
        print("child owns bike")
p1=Parent()  #create obejct and refered any varaibel 
p1.house()
p1.car() #call the values 
c1=Child()
c1.bike()
        
```

```python
#With Inheritance #
class Parent: #create class 
    def house(self):
        print("parent own house")
    def car(self):
        print("parent own car")
class Child(Parent): # inheritance calling Panrent class for reusing 
    def bike(self):
        print("child owns bike")

c1=Child() #creating objetct for child 

## Trying to access parent method from child method using inheritence ##
c1.house()
c1.car()
### ## Trying to access child method ##
c1.bike()
#note not cerated any obejct fro parent but still able to call parnet using child() obejct method with help of inheritance 
```

```python
# Inheritance exmaple-2#
class Bank:

    def __init__(self):
        self.__balance = 1000

    def show_balance(self):
        print("Available Balance:", self.__balance)

    def deposit(self, amount):
        self.__balance = amount + self.__balance
        print("Deposited:", amount)


class SavingsAccount(Bank):

    def interest(self):
        print("Savings account interest: 5%")


class CurrentAccount(Bank):

    def business(self):
        print("Business transactions enabled")


# Savings Account  # for saving account im using Bank() class using inheritance 
s1 = SavingsAccount()

s1.show_balance()
s1.deposit(500)
s1.show_balance()
s1.interest()


print("----------------")

# Current Account # # for current account im using Bank() class using inheritance 
c1 = CurrentAccount()

c1.show_balance()
c1.deposit(2000)
c1.show_balance()
c1.business()
```

```python
class Parent:
    def display(self):
        print("this is parent class")
class Child(Parent): #Inheritance 
    def display(self):
        print("this is child class")
c1 = Child()
c1.display()
## Child overrides the parent method , becuse both method names same (diplay)
```

## super method ##

```python
class Parent:
    def display(self):
        print("this is parent class")
class Child(Parent): #Inheritance 
    def display(self):
        super().display()
        print("this is child class")
c1 = Child()
c1.display()
```

```python
# inheritance with super init 
class Bank:

    def __init__(self):
        self.balance = 1000
class SavingsAccount(Bank):
    def __init__(self):
        super().__init__() #while using constuctors 
        self.account_type = "Saving Account"
        
s1 = SavingsAccount() #created object only for saving account but refered Bank using super inheritence 
print(s1.account_type)
print(s1.balance)
```

## Polymorphism
## same method name we can perform multiple operations 

```python
class Payment:

    def pay(self, amount):
        print("Processing payment")


class GooglePay(Payment):

    def pay(self, amount):
        print("Paid", amount, "through Google Pay")


class PhonePe(Payment):

    def pay(self, amount):
        print("Paid", amount, "through PhonePe")


class CreditCard(Payment):

    def pay(self, amount):
        print("Paid", amount, "through Credit Card")


class CashOnDelivery(Payment):

    def pay(self, amount):
        print("Cash will be collected on delivery:", amount)


# Customer selects payment method

#g = GooglePay()
p = PhonePe()
# c = CreditCard()
# cod = CashOnDelivery()

#g.pay(5000)
p.pay(3000)
# c.pay(10000)
# cod.pay(2000)
```

## Abstraction 
### Abstraction class define What should be done and child class defines how is should be done 

```python
from abc import ABC, abstractmethod


class ATM(ABC):  #for understanding consider parent 

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def check_balance(self):
        pass


class SBI(ATM):

    def __init__(self):
        self.balance = 10000

    def withdraw(self, amount):
        self.balance -= amount
        print("Withdrawn:", amount)

    def check_balance(self):
        print("Available Balance:", self.balance)
        
class HDFC(ATM):

    def __init__(self):
        self.balance = 20000

    def withdraw(self, amount):
        self.balance -= amount
        print("Withdrawn:", amount)

    def check_balance(self):
        print("Available Balance:", self.balance)

# Creating object
a = SBI() # #create object and refereb a to call the values form class 
a.withdraw(2000)
a.check_balance()

b = HDFC() #create object and refereb b to call the values form class 
b.withdraw(2000)
b.check_balance()

#abstract method class declared two mandatory methods 
#



```
