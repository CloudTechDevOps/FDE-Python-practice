```python
# with oops example for car1 and car2
class Car:
    def __init__(self, color, mileage, fuel):  #constructor of class Car
        self.color = color
        self.mileage = mileage
        self.fuel = fuel

    def start(self):
        print(f"Car is starting with color {self.color}, mileage {self.mileage} and fuel type {self.fuel}")

    def stop(self):
        print(f"Car is stopping with color {self.color}, mileage {self.mileage} and fuel type {self.fuel}")

    def move(self):
        print(f"Car is moving with color {self.color}, mileage {self.mileage} and fuel type {self.fuel}")
```

```python
car1 = Car("red", 15, "petrol")
car2 = Car("black", 10, "diesel")
car3 = Car("white", 12, "electric")
car1.start()
car1.stop()
car1.move()
car2.start()
car2.stop()
car2.move()
car3.start()
car3.stop()

```
