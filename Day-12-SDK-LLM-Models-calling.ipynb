## What is SDK
### SDK is caollection  of tool and libraries and functions provided by a company to help developres use to interact 
### SDK stands fro Software Development Kit
### Suppose we want to communicate with Gemini AI from Python. We don't need to develop all the communication logic from scratch.Google provides a ready-made SDK called Google GenAI, We install that SDK using pip, import it into Python, create a client and call ready-made methods.That's the main purpose of an SDK: making developers' work easier by providing ready-made tools and functions.
### Google provides a Python SDK called google-genai. -->pip install google-genai --> from google import genai
### OpenAI provides an official Python SDK called openai. -->pip install openai --> from openai import OpenAI
### Groq provides a Python SDK called groq. --> pip install groq --> from groq import Groq

```python
# pip install google-genai
from google import genai

client = genai.Client(api_key="") #api_key is the pre-defined paramater 

interaction = client.interactions.create(
    model="gemini-3.5-flash",
    input="current weather please in hyderbad?"
)

print(interaction.output_text)
```

## with functions

```python
from google import genai

client = genai.Client(api_key="update your key") #api_key is the pre-defined paramater 
def ask_gemini(prompt):
  interaction = client.interactions.create(
    model="gemini-3.5-flash",
    input= prompt
)
  return interaction.output_text
result=ask_gemini("what is python in one line?") #function call argument passing to paramter 
print(result)
```

```python
result1 = ask_gemini("what is python in one line?")
result2 = ask_gemini("what is java in one line?")
result3 = ask_gemini("what is aws in one line?")
print(result1)
print(result2)
print(result3)
```

## Call groq model

```python

from openai import OpenAI
import os
client = OpenAI(
    api_key=(api_key= "key")
    base_url="https://api.groq.com/openai/v1",
)

response = client.responses.create(
    input="Explain the importance of fast language models",
    model="openai/gpt-oss-20b",
)
print(response.output_text)
```
