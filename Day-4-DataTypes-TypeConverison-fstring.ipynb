## Data Types
### int,float,complex,bool,str,list,tuple,range,dic,set

```python
#int example 
age =25
count =10
temp=-10
print(type(temp).__name__)
```

```python
#float example  represent fractional component 
number = 25.0
print(type(number))
```

```python
# complex contains imaginary component 
number = 2+1j
print (type(number))
```

```python
# Boolean (bool) python has two boolean values Truth or False 
is_active = True
is_completes = False 
print (type(is_active))
```

```python
# String  it represnt text inside the quotes 
a="FDE"
b= "-45"
c= "55"
print(type(b))
```

```python
# list is an order and mutable collection and allow duplicate values []
number = [10, "veera", "-1"] 
print (number)
print (type(number))
```

```python
# tuple an order collection and immutable ()
number = (10, "veera", "-1") 
print (number)
print (type(number))
```

```python
# Set mutable collection of unique element {}
number = {10, 20, 30, 30}
print (type(number))
```

```python
#Dictionary (dict) stores key-values pairs
student = {
    "name": "ravi",
    "age" : 25,
    "course" : "FDE"
}
print(student)
print(type(student))
```

```python
#None Type absence of the value
result = None 
print(type(result))
```

```python
# range sequesnce of integer 
number = range(5)
print(number)
print (list(number))
print (type(number))
```
