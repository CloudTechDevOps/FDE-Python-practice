# Day-1 Identifier-variablle 
### Identifier: A name used to identify a program element such as variables, functions, class etc..
### Variables: A name that refer to an object in memeory during execution 

## Rules to define a varaibel or Identifier 

#### Use letter and digits and underscore_
#### Never start identifier with digit
#### Do not use python keywords as identifers ex: if, for etc..
#### Do not use spaces hyphens -, in indentifier 
#### Better to use snake_case ex server_name, SERVER_NAME
#### Aviod using python built-in functions name as variable names
