## Python Exception Handling 
### try, except, else, finally , raise 

### alone try or exception block not allowed

```python
a= 10
b= 0
print (a/b)
```

```python
try:
    a= 10
    b= 0
    print (a/b)
except:
    print("cannot devide by zero")
```

```python
try:
    a= 10
    b= 0
    print (a/b)
except Exception as e:
    print(e)
    # e is varible that store exception object (error information )
```

```python
count = 1
while count <=3:
    try:
        a=10
        b=1
        print(a/b)
        count +=1
    except Exception as e:
        print(e)
        
       
```

```python
try:
    a = 10
    b = 0
    print (a/b)
except:
    print("cannot devide by zero")
finally:
    print("execution completed")
```

```python
try:
    a= int(input("enter number for a"))
    b= int(input("enter number for b"))
    result = a/b
    print(result)
except ZeroDivisionError:
    print("cannot divide by zero")
except ValueError:
    print("please enter correct number only")
except TypeError:
    print("invalide data type")
except Exception as e:
    print("unexpecred error:", e)
```

```python
try:
    a= int(input("enter number for a"))
    b= int(input("enter number for b"))
    print(a/b)

except (ValueError,ZeroDivisionError) as e:
    print("please enter correct number only", e)
else:
    print("else excecution")
finally:
    print("program exceution completed")

```

```python
def process_data(items, index, divisor):
    try:
        value = int(items[index])
        result = value / divisor
    except (IndexError, KeyError):
        print("Error: The requested item could not be found.")
    except ValueError:
        print("Error: The item could not be converted to a number.")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
    else:
        print(f"Success! The result is {result}.")

process_data(["10", "abc"], 1, 2)  # Triggers ValueError
process_data(["10", "20"], 5, 2)   # Triggers IndexError
```
