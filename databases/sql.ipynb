## SQLite is a lightweight, serverless, file-based relational database management system (RDBMS).

### In Python, SQLite is available through the built-in sqlite3 module.

### Important: You don't need to install a separate database server like MySQL or PostgreSQL to use SQLite.

| Step      | Method               | Purpose |
|
| 1 | `sqlite3.connect()` | Establish database connection |
| 2 | `conn.cursor()`     | Create cursor object |
| 3 | `cur.execute()`     | Execute SQL queries |
| 4 | `conn.commit()`     | Save changes |
| 5 | `cur.fetchall()`    | Retrieve records |
| 6 | `conn.close()`      | Close database connection |

sqlite3 is a built-in Python module.
It provides functionality to communicate with SQLite databases.
connect() is a function.
It establishes a connection with the database.
conn is a variable.
It stores the connection object.
Important: If students.db doesn't exist, SQLite automatically creates the database file.

```python
import sqlite3

# sqlite3 is a built-in Python module.
# It provides functionality to communicate with SQLite database
```

## Establish Database Connection

```python
conn = sqlite3.connect("students.db")

print("Database connected successfully")

# sqlite3 is a built-in Python module.
# It provides functionality to communicate with SQLite databases.
# connect() is a function.
# It establishes a connection with the database.
# conn is a variable.
# It stores the connection object.
# Important: If students.db doesn't exist, SQLite automatically creates the database file.
```

## Create Cursor

```python
cur = conn.cursor()

print("Cursor created successfully")

# cursor() is an instance method.
# It creates a cursor object.
# cur is a variable that stores the cursor object.
# We use the cursor to execute SQL queries.
```

## Create Table

```python
cur.execute("""
CREATE TABLE studentss (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    course TEXT
)
""")

conn.commit()

# execute() executes SQL statements.
# CREATE TABLE creates a table.
# commit() saves the changes permanently.

# Scenario 1: Without commit()
# cur.execute("INSERT INTO students VALUES (1, 'Veera', 30)")   

#if encounter an issue   

# What happens?
# - INSERT executes successfully.
# - Program encounters an error.
# - commit() never executes.
# - Connection closes.
# - Uncommitted transaction is rolled back.
# Result: Record is NOT saved.


# Scenario 2: With commit()
# cur.execute("INSERT INTO students VALUES (1, 'Veera', 30)")conn.commit() 

#if encounter as issue 

# What happens?
# - INSERT executes successfully.
# - commit() saves the record.
# - Program encounters an error.
# - Connection closes.
# Result: Records  saved.
```

## Insert Records

```python
cur.execute("""
INSERT INTO students (name, age, course)
VALUES ('Veera', 30, 'Python')
""")

conn.commit()

print("Record inserted successfully")
```

```python
query = """
INSERT INTO students (name, age, course)
VALUES (?, ?, ?)
"""

name = "Veera"
age = 30
course = "Python"

cur.execute(query, (name, age, course))

conn.commit()

print("Record inserted successfully")

# ? is called a placeholder in SQLite parameterized queries.

#It acts as a temporary position where SQLite inserts the actual value.
```

## Retrieve Records

```python
cur.execute("SELECT * FROM students")

records = cur.fetchall()

print(records)
#fetchall() retrieves all records returned by the SQL query.
```

## Closes the database connection and releases associated resources.

```python
conn.close()

print("Database connection closed")

#We close database connections to release resources, prevent unnecessary database locking, reduce security exposure, and maintain application performance."
```

## try exception examples

## The best practice is to use try, except, commit(), rollback(), and finally together when performing database write operations.

```python
import sqlite3

conn = sqlite3.connect("/invalid_folder/students.db")

print("Database connected successfully")
```

```python
import sqlite3

try:
    conn = sqlite3.connect("/invalid_folder/students.db")
    print("Database connected successfully")

except sqlite3.Error as e:  #sqlite3.Error is a predefined exception class provided by the SQLite module.
    print("Database Error:", e)
```

```python
import sqlite3

try:
    conn = sqlite3.connect("/invalid_folder/students.db")
    print("Database connected successfully")

except sqlite3.Error as e:
    print("Database Error:", e)

finally:
    print("Database connection closed")
```

```python
import sqlite3

try:
    conn = sqlite3.connect("/invalid_folder/students.db")
    print("Database connected successfully")

except sqlite3.Error as e:
    print("Database Error:", e)

finally:
    try:
        conn.close()
        print("Database connection closed")

    except:
        print("Database connection was not established")
```

```python
import sqlite3

try:
    conn = sqlite3.connect("students.db")
    cur = conn.cursor()

    cur.execute("""
        INSERT INTO students (name, age)
        VALUES (?, ?)
    """, ("Veera", 30))

    # Save changes
    conn.commit()

    print("Record inserted successfully")

except sqlite3.Error as e:

    # Cancel uncommitted changes
    conn.rollback()

    print("Database Error:", e)

finally:
    if 'conn' in locals(): #	Checks whether conn exists Before closing the database connection, check whether the connection variable was successfully created.
        conn.close()

    print("Connection closed")
```

## CRUD represents four important database operations.
### CRUD stands for Create, Read, Update, and Delete.
### These are the four basic operations performed on a database.

```python
import sqlite3

conn = sqlite3.connect("students.db")
cur = conn.cursor()

# CREATE
cur.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER
)
""")

# INSERT
cur.execute(
    "INSERT INTO students (name, age) VALUES (?, ?)",
    ("Veera", 30)
)

# READ
cur.execute("SELECT * FROM students")
print("Students:", cur.fetchall())

# UPDATE
cur.execute(
    "UPDATE students SET age = ? WHERE name = ?",
    (31, "Veera")
)

# DELETE
cur.execute(
    "DELETE FROM students WHERE name = ?",
    ("Veera",)
)

conn.commit()
conn.close()
```

```python
| Operation | SQL Command | Purpose |

| Create | INSERT | Insert records |
| Read   | SELECT | Retrieve records |
| Update | UPDATE | Modify records |
| Delete | DELETE | Remove records |
```
